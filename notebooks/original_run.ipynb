```python
from transformers import pipeline
import pandas as pd
```

```python
CANDIDATE_LABELS = ["phishing", "legitimate"]

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=0
)

classifier_roberta = pipeline(
    "zero-shot-classification",
    model="roberta-large-mnli",
    device=0
)

classifier_deberta = pipeline(
    "zero-shot-classification",
    model="MoritzLaurer/deberta-v3-large-zeroshot-v2.0",
    device=0
)

classifier_distil = pipeline(
    "zero-shot-classification",
    model="valhalla/distilbart-mnli-12-3",
    device=0
)

models = {
    "bart": classifier,
    "roberta": classifier_roberta,
    "deberta": classifier_deberta,
    "distilbart": classifier_distil
}

print("All models loaded successfully.")
```

```python
print("Total groups:", len(paraphrase_groups))
print("Phishing:", sum(1 for g in paraphrase_groups if g["label"] == "phishing"))
print("Legitimate:", sum(1 for g in paraphrase_groups if g["label"] == "legitimate"))
```

```python
all_results = []

for model_name, clf in models.items():
    print(f"\nRunning model: {model_name}\n")

    for group in paraphrase_groups:
        group_predictions = []

        for i, variant in enumerate(group["variants"]):
            output = clf(
                variant,
                candidate_labels=CANDIDATE_LABELS,
                multi_label=False
            )

            if isinstance(output, list):
                output = output[0]

            predicted_label = output["labels"][0]
            confidence = output["scores"][0]

            all_results.append({
                "model": model_name,
                "group_id": group["group_id"],
                "true_label": group["label"],
                "paraphrase_type": group["paraphrase_type"],
                "variant_index": i + 1,
                "variant_text": variant,
                "predicted_label": predicted_label,
                "confidence": round(confidence, 4)
            })

            group_predictions.append(predicted_label)

        print(f"Group {group['group_id']} → {group_predictions}")

df_all = pd.DataFrame(all_results)

print("\nAll models completed.")
print(df_all.shape)
```

```python
def compute_consistency_score(df, filter_col=None, filter_val=None):
    if filter_col and filter_val:
        df = df[df[filter_col] == filter_val]

    grouped = df.groupby(["model", "group_id"])["predicted_label"].apply(list)

    consistent = grouped.apply(lambda preds: len(set(preds)) == 1)

    cs = consistent.mean()
    n = len(consistent)

    return round(cs, 4), n
```

```python
df_all["correct"] = df_all["predicted_label"] == df_all["true_label"]

accuracy_by_model = df_all.groupby("model")["correct"].mean().round(4)

print("Accuracy by model:")
print(accuracy_by_model)
```

```python
print("=" * 65)
print("TABLE 1: Overall Results (All Models)")
print("=" * 65)
print(f"{'Model':<20} {'CS':>6} {'Accuracy':>10} {'Groups':>8}")
print("-" * 65)

for model in df_all["model"].unique():
    df_m = df_all[df_all["model"] == model]

    cs, n = compute_consistency_score(df_m)
    acc = df_m["correct"].mean()

    print(f"{model:<20} {cs:>6} {round(acc,4):>10} {n:>8}")

print(f"{'Random baseline':<20} {0.25:>6} {0.50:>10} {60:>8}")
```

```python
print("\n" + "=" * 65)
print("TABLE 2: CS by Paraphrase Type")
print("=" * 65)

for model in df_all["model"].unique():
    print(f"\nModel: {model}")

    df_m = df_all[df_all["model"] == model]

    for ptype in ["synonym", "syntactic", "manual"]:
        cs, n = compute_consistency_score(df_m, "paraphrase_type", ptype)
        print(f"{ptype:<20} CS={cs}  n={n}")
```

```python
print("\n" + "=" * 65)
print("TABLE 3: CS by Label")
print("=" * 65)

for model in df_all["model"].unique():
    print(f"\nModel: {model}")

    df_m = df_all[df_all["model"] == model]

    for label in ["phishing", "legitimate"]:
        cs, n = compute_consistency_score(df_m, "true_label", label)
        print(f"{label:<20} CS={cs}  n={n}")
```
